```python
import pandas as pd

# （1）通过字典创建 Series s1
data1 = {'TV': 999.00, 'Phone': 899.00, 'PC': 1999.00, 'VR': 99.00}
s1 = pd.Series(data1)
print("s1:")
print(s1)

# （2）通过字典创建 Series s2，并指定索引
data2 = {'TV': 999.00, 'Phone': 899.00, 'PC': 1999.00, 'VR': 99.00}
s2 = pd.Series(data2, index=['TV', 'Phone', 'VR', 'AI'])
print("\ns2:")
print(s2)

# （3）s1 + s2 相加
print("\ns1 + s2:")
print(s1 + s2)
```

```python
import pandas as pd

# （1）使用 DataFrame 保存数据
data = {
    '商品': ['电脑', '平板', '手机', '显示器'],
    '销售额': [12000, 8500, 15000, 6000],
    '销量': [30, 50, 75, 20],
    '库存': [15, 30, 40, 10]
}
df = pd.DataFrame(data)
print("原始数据：")
print(df)

# （2）添加“单价”列（单价 = 销售额 / 销量）
df['单价'] = df['销售额'] / df['销量']
print("\n添加单价列后：")
print(df)

# （3）输出“商品”和“库存”两列
print("\n商品和库存列：")
print(df[['商品', '库存']])

# （4）输出电脑和手机的销售额和销量
print("\n电脑和手机的销售额与销量：")
result = df[df['商品'].isin(['电脑', '手机'])][['商品', '销售额', '销量']]
print(result)
```

```python
import pandas as pd

# 构建多层列索引：(组别, 学生)
columns = pd.MultiIndex.from_tuples(
    [('A组', '小红'), ('A组', '小明'), ('B组', '小王'), ('C组', '小绿')],
    names=['组别', '学生']
)

# 构建多层行索引：(学期, 科目)
index = pd.MultiIndex.from_tuples(
    [('第一学期', '语文'), ('第一学期', '数学'), ('第一学期', '英语'),
     ('第二学期', '语文'), ('第二学期', '数学')],
    names=['学期', '科目']
)

# 成绩数据
data = [
    [90, 80, 70, 60],
    [90, 80, 70, 60],
    [90, 80, 70, 60],
    [90, 80, 70, 60],
    [90, 80, 70, 60]
]

# （1）创建并输出 DataFrame
df = pd.DataFrame(data, index=index, columns=columns)
print("（1）完整数据：")
print(df)

# （2）输出 A 组所有学生的成绩
print("\n（2）A组所有学生成绩：")
print(df['A组'])

# （3）输出第一学期所有学生所有科目的成绩
print("\n（3）第一学期所有成绩：")
print(df.loc['第一学期'])

# （4）输出小红第一学期的成绩
print("\n（4）小红第一学期成绩：")
print(df.loc[('第一学期', slice(None)), ('A组', '小红')])
```

```python
import pandas as pd
import numpy as np

# ===================== 加载数据集 =====================
# 确保 scientists.csv 和当前 .py 文件在同一文件夹
scientists = pd.read_csv('scientists.csv')

# ===================== 第一部分：Series 统计 =====================
print("="*30 + " 第一部分 " + "="*30)
# （1）获取 Age 列的数据保存为 age_series
age_series = scientists['Age']
print("（1）年龄 Series：")
print(age_series)

# （2）计算年龄的平均值、最大值、最小值、标准差
print("\n（2）年龄统计值：")
print(f"平均值：{age_series.mean():.2f}")
print(f"最大值：{age_series.max()}")
print(f"最小值：{age_series.min()}")
print(f"标准差：{age_series.std():.2f}")

# （3）获取职业列保存为 occupation_series，并显示各类科学家人数
occupation_series = scientists['Occupation']
print("\n（3）职业分布（各类科学家人数）：")
print(occupation_series.value_counts())

# （4）统计 Born 这一列非空元素的个数
print("\n（4）Born 列非空元素个数：")
print(scientists['Born'].count())

# （5）显示 age_series 和 occupation_series 的各种统计值
print("\n（5）age_series 统计描述：")
print(age_series.describe())
print("\noccupation_series 统计描述：")
print(occupation_series.describe())

# ===================== 第二部分：DataFrame 统计 =====================
print("\n" + "="*30 + " 第二部分 " + "="*30)
# （1）获取最大值、最小值、每列非空元素个数
print("（1）数值列最大值：")
print(scientists.max(numeric_only=True))  # 仅数值列
print("\n数值列最小值：")
print(scientists.min(numeric_only=True))
print("\n每列非空元素个数：")
print(scientists.count())

# （2）分别获取数值型列和非数值型列的统计信息
print("\n（2）数值型列统计信息：")
print(scientists.describe())  # 默认只显示数值列
print("\n非数值型列统计信息：")
print(scientists.describe(include=[np.object_]))  # 仅显示文本/类别列

# ===================== 第三部分：数据筛选 =====================
print("\n" + "="*30 + " 第三部分 " + "="*30)
# （1）从 age_series 筛选年龄大于平均值的数据
age_mean = age_series.mean()
age_above_mean = age_series[age_series > age_mean]
print(f"（1）年龄大于平均值（{age_mean:.2f}）的数据：")
print(age_above_mean)

# （2）获取 scientists 中 Age 大于平均值的科学家完整信息
scientists_above_mean = scientists[scientists['Age'] > age_mean]
print("\n（2）年龄大于平均值的科学家完整信息：")
print(scientists_above_mean)
```

```python
import pandas as pd

# 1. 加载 student.xlsx 数据（确保文件和代码在同一文件夹，或使用绝对路径）
df = pd.read_excel('student.xlsx', engine='openpyxl')

# 2. 查看原始列名（确认科目列，方便计算总分）
print("原始列名：")
print(df.columns.tolist())

# 3. 增加“总分”列（假设科目为 ['语文', '数学', '英语']，请根据实际列名修改）
# 示例：df['总分'] = df['语文'] + df['数学'] + df['英语']
# 自动求和所有数值列（更通用的写法）
df['总分'] = df.select_dtypes(include='number').sum(axis=1)

# 4. 按照总分从高到低排序（ascending=False 表示降序）
df_sorted = df.sort_values(by='总分', ascending=False, ignore_index=True)

# 5. 输出结果
print("\n添加总分并排序后的数据：")
print(df_sorted)

# 可选：将结果保存到新 Excel 文件
df_sorted.to_excel('student_总分排序.xlsx', index=False)
print("\n✅ 已将排序后的数据保存为：student_总分排序.xlsx")
```
