```python
import numpy as np

sales = np.array([
    [50, 45, 60, 55, 70, 90, 85],  # 苹果
    [30, 25, 35, 40, 45, 60, 55],  # 香蕉
    [40, 35, 45, 50, 55, 80, 75],  # 橙子
    [20, 15, 25, 30, 35, 50, 45]   # 葡萄
])


fruits = ['苹果', '香蕉', '橙子', '葡萄']
weekdays = ['周一', '周二', '周三', '周四', '周五', '周六', '周日']

q1 = sales[0, 2]
print(f"1. 苹果周三销量: {q1}")


q2 = sales[0, [2, 4]]
print(f"2. 苹果周三、周五销量: {q2}")


q3 = sales[3, 5:7]
print(f"3. 葡萄周末销量: {q3}")


q4 = sales[:, 4]
print(f"4. 周五所有水果销量: {q4}")


weekend_sales = sales[:, 5:7]

mask = np.any(weekend_sales > 60, axis=1)

q5 = np.array(fruits)[mask]
print(f"5. 周末销量大于60的水果: {q5}")
```
