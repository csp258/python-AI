```python
import numpy as np


scores = np.array([
    [85, 92, 78],  # 学生A
    [90, 88, 95],  # 学生B
    [76, 85, 82],  # 学生C
    [92, 79, 88],  # 学生D
    [88, 95, 90]   # 学生E
])

students = ['学生A', '学生B', '学生C', '学生D', '学生E']
subjects = ['语文', '数学', '英语']


extra = np.array([2, 3, 1, 4, 2]).reshape(-1, 1)  # 转为列向量，方便广播
scores_with_extra = scores + extra
```

```python

total_scores = scores_with_extra.sum(axis=1)


for i, student in enumerate(students):
    print(f"{student}：{total_scores[i]}分")
```

```python

rank_idx = np.argsort(total_scores)[::-1]


sorted_students = [students[i] for i in rank_idx]
sorted_totals = total_scores[rank_idx]


print("\n按总分从高到低排序：")
for student, total in zip(sorted_students, sorted_totals):
    print(f"{student}：{total}分")
```

```python

scores_transposed = scores_with_extra.T

print("\n转置后的数组（行：科目，列：学生A-E）：")
print(scores_transposed)
```

```python

chinese_scores, other_scores = np.hsplit(scores_with_extra, [1])

print("\n语文成绩：")
print(chinese_scores)
print("\n其他科目成绩（数学、英语）：")
print(other_scores)
```
