```python
from matplotlib.axes import Axes
import matplotlib.pyplot as plt


#展示中文
plt.rcParams['font.sans-serif'] = ['SimHei'] #设置中文字体为黑体

#创建子图
#figure 图表对象 axes 子图对象
figure,axes = plt.subplots(nrows = 1,ncols = 2,figsize = (20,6)) #nrows 行数 ncols 列数 figsize 图表布大小

#图一:柱状图(世界石油储备)
countries = ['中国','美国','俄罗斯','日本','韩国','伊拉克','伊朗']
values = [35,23,18,15,12,10,9]


axes1:Axes = axes[0]
axes1.bar(countries,values)
axes1.set_title('世界石油储备',fontsize = 18)#设置标题 字体大小
axes1.set_xlabel('国家',fontsize = 14)#设置x轴标签 字体大小
axes1.set_ylabel('石油储备(亿桶)',fontsize = 14)#设置y轴标签 字体大小
axes1.grid(linestyle = '--',alpha = 0.3)#添加网格线



#图二:饼状图(世界人口) --->擅长比例分析
countries2 = ['中国','美国','俄罗斯','日本','韩国','英国','其他']
values = [14.51,14.09,1.02,1.01,1.26,1.14,1.07]
axes2:Axes = axes[1]
axes2.set_title('世界人口比例',fontsize = 18) #设置标题 字体大小
axes2.pie(values,labels = countries2,autopct='%1.1f%%');#autopct 自动添加百分比标签
axes2.legend(loc = 'lower center',ncol = 5,bbox_to_anchor = (0.5,-0.15))#添加图例  loc 图例位置 ncol 图例列数 bbox_to_anchor 图例位置偏移量
#(0.5控制水平方位,-0.15控制垂直方位)














```
